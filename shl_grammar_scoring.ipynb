```python
import os
import gc
import numpy as np
import torch
import librosa
from transformers import AutoFeatureExtractor, AutoModel

# --------------------------------------------------
# Paths
# --------------------------------------------------
TRAIN_DIR = "/content/audio_train"
TEST_DIR = "/content/audio_test"
CHECKPOINT_DIR = "/content/drive/MyDrive/SHL_Grammar/checkpoints"

os.makedirs(CHECKPOINT_DIR, exist_ok=True)

# --------------------------------------------------
# Model
# --------------------------------------------------
MODEL_NAME = "microsoft/wavlm-base-plus"
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)

processor = AutoFeatureExtractor.from_pretrained(MODEL_NAME)
wavlm = AutoModel.from_pretrained(MODEL_NAME).to(device)
wavlm.eval()

# --------------------------------------------------
# Function: one audio -> 768-dim WavLM embedding
# --------------------------------------------------
def get_embedding(path):
    y, sr = librosa.load(path, sr=16000, mono=True)

    inputs = processor(
        y,
        sampling_rate=16000,
        return_tensors="pt"
    )

    inputs = {
        k: v.to(device)
        for k, v in inputs.items()
    }

    with torch.no_grad():
        outputs = wavlm(**inputs)

    # Mean pooling over time
    emb = outputs.last_hidden_state.mean(dim=1)

    return emb.squeeze(0).cpu().numpy().astype(np.float32)


# --------------------------------------------------
# Checkpointed extraction
# --------------------------------------------------
def extract_dataset(audio_dir, output_name, prefix):

    files = sorted([
        f for f in os.listdir(audio_dir)
        if f.lower().endswith(".wav")
    ])

    checkpoint_file = os.path.join(
        CHECKPOINT_DIR,
        f"{prefix}_checkpoint.npy"
    )

    final_file = os.path.join(
        CHECKPOINT_DIR,
        output_name
    )

    # Resume from checkpoint if available
    if os.path.exists(checkpoint_file):
        embeddings = list(np.load(checkpoint_file))
        start_idx = len(embeddings)

        print(
            f"\nResuming {prefix}: "
            f"{start_idx}/{len(files)} already completed"
        )
    else:
        embeddings = []
        start_idx = 0

        print(
            f"\nStarting {prefix}: "
            f"{len(files)} files"
        )

    for i in range(start_idx, len(files)):

        filename = files[i]
        path = os.path.join(audio_dir, filename)

        try:
            emb = get_embedding(path)
            embeddings.append(emb)

        except Exception as e:
            print(f"\nERROR: {filename}")
            print(e)
            raise

        # Progress
        if (i + 1) % 10 == 0 or i == start_idx:
            print(
                f"{prefix}: {i + 1}/{len(files)}"
            )

        # Save checkpoint every 25 files
        if (i + 1) % 25 == 0:
            arr = np.stack(embeddings)

            np.save(checkpoint_file, arr)

            print(
                f"  💾 Checkpoint saved: "
                f"{arr.shape}"
            )

        # Keep memory clean
        if (i + 1) % 50 == 0:
            gc.collect()
            torch.cuda.empty_cache()

    # Final array
    embeddings = np.stack(embeddings).astype(np.float32)

    np.save(final_file, embeddings)

    # Keep checkpoint too for safety
    np.save(checkpoint_file, embeddings)

    print(
        f"\n✅ {prefix} complete:"
        f" {embeddings.shape}"
    )
    print("Saved:", final_file)

    return embeddings


# --------------------------------------------------
# TRAIN
# --------------------------------------------------
train_wavlm = extract_dataset(
    TRAIN_DIR,
    "train_wavlm.npy",
    "train_wavlm"
)

# --------------------------------------------------
# TEST
# --------------------------------------------------
test_wavlm = extract_dataset(
    TEST_DIR,
    "test_wavlm.npy",
    "test_wavlm"
)

print("\n===================================")
print("FINAL EMBEDDING SHAPES")
print("Train:", train_wavlm.shape)
print("Test :", test_wavlm.shape)
print("===================================")
```

```python
import os
import numpy as np
import pandas as pd
import librosa

base = "/content/drive/MyDrive/SHL_Grammar"
checkpoint_dir = os.path.join(base, "checkpoints")

train = pd.read_csv(os.path.join(base, "train.csv"))
test = pd.read_csv(os.path.join(base, "test.csv"))

def extract_features(audio_path):
    y, sr = librosa.load(
        audio_path,
        sr=16000,
        mono=True
    )

    features = {}

    # MFCC
    mfcc = librosa.feature.mfcc(
        y=y,
        sr=sr,
        n_mfcc=13
    )

    for i in range(13):
        features[f"mfcc_{i}_mean"] = np.mean(mfcc[i])
        features[f"mfcc_{i}_std"] = np.std(mfcc[i])

    # RMS energy
    features["rms"] = np.mean(
        librosa.feature.rms(y=y)
    )

    # Zero crossing rate
    features["zero_crossing_rate"] = np.mean(
        librosa.feature.zero_crossing_rate(y)
    )

    # Spectral centroid
    features["spectral_centroid"] = np.mean(
        librosa.feature.spectral_centroid(
            y=y,
            sr=sr
        )
    )

    # Spectral bandwidth
    features["spectral_bandwidth"] = np.mean(
        librosa.feature.spectral_bandwidth(
            y=y,
            sr=sr
        )
    )

    return features


def extract_dataset_features(audio_dir, filenames):
    rows = []

    for i, filename in enumerate(filenames):

        path = os.path.join(
            audio_dir,
            filename
        )

        features = extract_features(path)
        rows.append(features)

        if (i + 1) % 50 == 0 or i == 0:
            print(
                f"{i + 1}/{len(filenames)}"
            )

    return pd.DataFrame(rows)


# IMPORTANT:
# Use CSV order so the features align directly
# with the labels and WavLM embeddings.

print("Extracting TRAIN handcrafted features...")

train_handcrafted = extract_dataset_features(
    "/content/audio_train",
    train["filename"].tolist()
)

print("\nExtracting TEST handcrafted features...")

test_handcrafted = extract_dataset_features(
    "/content/audio_test",
    test["filename"].tolist()
)

# Save permanently
train_handcrafted.to_csv(
    os.path.join(
        checkpoint_dir,
        "train_handcrafted.csv"
    ),
    index=False
)

test_handcrafted.to_csv(
    os.path.join(
        checkpoint_dir,
        "test_handcrafted.csv"
    ),
    index=False
)

print("\n================================")
print("TRAIN:", train_handcrafted.shape)
print("TEST :", test_handcrafted.shape)
print("================================")

print("\nMissing values:")
print(
    "Train:",
    train_handcrafted.isna().sum().sum()
)

print(
    "Test :",
    test_handcrafted.isna().sum().sum()
)

print("\nSaved handcrafted features to Drive.")
```

```python
import os
import gc
import numpy as np
import torch
import librosa
from transformers import AutoFeatureExtractor, AutoModel

# --------------------------------------------------
# Paths
# --------------------------------------------------
TRAIN_DIR = "/content/audio_train"
TEST_DIR = "/content/audio_test"

CHECKPOINT_DIR = "/content/drive/MyDrive/SHL_Grammar/checkpoints"
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

# --------------------------------------------------
# WavLM
# --------------------------------------------------
MODEL_NAME = "microsoft/wavlm-base-plus"

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

processor = AutoFeatureExtractor.from_pretrained(
    MODEL_NAME
)

wavlm = AutoModel.from_pretrained(
    MODEL_NAME
).to(device)

wavlm.eval()

print("Device:", device)


# --------------------------------------------------
# Extract one 4-region temporal representation
# --------------------------------------------------
def get_temporal_embedding(path):

    y, sr = librosa.load(
        path,
        sr=16000,
        mono=True
    )

    n = len(y)

    # Four chronological regions.
    # 10% overlap around boundaries to preserve context.
    boundaries = np.linspace(0, n, 5).astype(int)

    chunks = []

    for i in range(4):

        start = boundaries[i]
        end = boundaries[i + 1]

        # Add 10% of one segment as context
        segment_length = end - start
        context = int(0.10 * segment_length)

        start = max(0, start - context)
        end = min(n, end + context)

        chunk = y[start:end]

        inputs = processor(
            chunk,
            sampling_rate=16000,
            return_tensors="pt"
        )

        inputs = {
            k: v.to(device)
            for k, v in inputs.items()
        }

        with torch.no_grad():
            outputs = wavlm(**inputs)

        # Mean pooling within this temporal region
        emb = outputs.last_hidden_state.mean(dim=1)

        chunks.append(
            emb.squeeze(0).cpu().numpy()
        )

        del inputs, outputs, emb

    # 4 × 768 = 3072
    return np.concatenate(chunks).astype(np.float32)


# --------------------------------------------------
# Checkpointed dataset extraction
# --------------------------------------------------
def extract_temporal_dataset(
    audio_dir,
    filenames,
    prefix
):

    checkpoint_file = os.path.join(
        CHECKPOINT_DIR,
        f"{prefix}_checkpoint.npy"
    )

    final_file = os.path.join(
        CHECKPOINT_DIR,
        f"{prefix}.npy"
    )

    # Resume if checkpoint exists
    if os.path.exists(checkpoint_file):

        embeddings = list(
            np.load(checkpoint_file)
        )

        start_idx = len(embeddings)

        print(
            f"\nResuming {prefix}: "
            f"{start_idx}/{len(filenames)}"
        )

    else:

        embeddings = []
        start_idx = 0

        print(
            f"\nStarting {prefix}: "
            f"{len(filenames)} files"
        )

    for i in range(start_idx, len(filenames)):

        filename = filenames[i]

        path = os.path.join(
            audio_dir,
            filename
        )

        embedding = get_temporal_embedding(path)

        embeddings.append(embedding)

        if (i + 1) % 5 == 0 or i == start_idx:

            print(
                f"{prefix}: "
                f"{i + 1}/{len(filenames)}"
            )

        # Save every 10 files
        if (i + 1) % 10 == 0:

            arr = np.stack(embeddings)

            np.save(
                checkpoint_file,
                arr
            )

            print(
                f"  💾 checkpoint: {arr.shape}"
            )

        if (i + 1) % 25 == 0:

            gc.collect()
            torch.cuda.empty_cache()

    # Final array
    embeddings = np.stack(
        embeddings
    ).astype(np.float32)

    np.save(
        final_file,
        embeddings
    )

    np.save(
        checkpoint_file,
        embeddings
    )

    print(
        f"\n✅ {prefix}: "
        f"{embeddings.shape}"
    )

    return embeddings


# --------------------------------------------------
# TRAIN
# --------------------------------------------------
train_temporal = extract_temporal_dataset(
    TRAIN_DIR,
    train["filename"].tolist(),
    "train_wavlm_temporal"
)

# --------------------------------------------------
# TEST
# --------------------------------------------------
test_temporal = extract_temporal_dataset(
    TEST_DIR,
    test["filename"].tolist(),
    "test_wavlm_temporal"
)

print("\n==============================")
print("FINAL")
print("Train:", train_temporal.shape)
print("Test :", test_temporal.shape)
print("==============================")
```

```python
import numpy as np
import pandas as pd

from sklearn.metrics import roc_auc_score
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# Original 798 features
X_train_domain = X_combined
X_test_domain = X_test_combined

X_domain = np.vstack([
    X_train_domain,
    X_test_domain
])

domain_y = np.concatenate([
    np.zeros(len(X_train_domain)),
    np.ones(len(X_test_domain))
])

# Feature names
feature_names = (
    [f"wavlm_{i}" for i in range(768)]
    + list(train_handcrafted.columns)
)

print("Total features:", len(feature_names))

# Evaluate each feature individually
results = []

for j, name in enumerate(feature_names):

    values = X_domain[:, j].reshape(-1, 1)

    model = make_pipeline(
        StandardScaler(),
        LogisticRegression(
            max_iter=1000,
            random_state=42
        )
    )

    cv = StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    )

    pred = cross_val_predict(
        model,
        values,
        domain_y,
        cv=cv,
        method="predict_proba"
    )[:, 1]

    auc = roc_auc_score(
        domain_y,
        pred
    )

    # Distance from random guessing
    shift_strength = abs(auc - 0.5)

    results.append({
        "feature": name,
        "auc": auc,
        "shift_strength": shift_strength
    })

shift_df = pd.DataFrame(results)

shift_df = shift_df.sort_values(
    "shift_strength",
    ascending=False
).reset_index(drop=True)

print("\nMost shifted features:")
print(
    shift_df.head(30).to_string(index=False)
)

print("\nLeast shifted features:")
print(
    shift_df.tail(10).to_string(index=False)
)
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error, roc_auc_score
from sklearn.ensemble import RandomForestClassifier

# ==================================================
# LOAD DATA
# ==================================================

train_wavlm = np.load(
    "/content/drive/MyDrive/SHL_Grammar/checkpoints/train_wavlm_aligned.npy"
)

test_wavlm = np.load(
    "/content/drive/MyDrive/SHL_Grammar/checkpoints/test_wavlm_aligned.npy"
)

train_hand = pd.read_csv(
    "/content/drive/MyDrive/SHL_Grammar/checkpoints/train_handcrafted.csv"
)

test_hand = pd.read_csv(
    "/content/drive/MyDrive/SHL_Grammar/checkpoints/test_handcrafted.csv"
)

hand_cols = [c for c in train_hand.columns if c != "filename"]

train_hand_X = train_hand[hand_cols].values
test_hand_X = test_hand[hand_cols].values

y = train["label"].values

# ==================================================
# 1. TARGET CORRELATION
# ==================================================

# Pearson correlation of each WavLM dimension with target
target_corr = np.array([
    np.corrcoef(train_wavlm[:, i], y)[0, 1]
    for i in range(train_wavlm.shape[1])
])

target_strength = np.abs(target_corr)

print("WavLM target correlation")
print("=========================")
print("Highest |correlation|:", round(target_strength.max(), 4))
print("Median |correlation| :", round(np.median(target_strength), 4))

# ==================================================
# 2. DOMAIN SHIFT SCORE
# ==================================================

# Train/test domain labels
X_domain = np.vstack([
    train_wavlm,
    test_wavlm
])

domain_y = np.concatenate([
    np.zeros(len(train_wavlm)),
    np.ones(len(test_wavlm))
])

# Scale domain features
domain_scaler = StandardScaler()
X_domain_scaled = domain_scaler.fit_transform(X_domain)

# One feature at a time:
# AUC close to 0.5 = stable
# AUC far from 0.5 = shifted

domain_auc = np.zeros(train_wavlm.shape[1])

for i in range(train_wavlm.shape[1]):
    try:
        domain_auc[i] = roc_auc_score(
            domain_y,
            X_domain_scaled[:, i]
        )
    except:
        domain_auc[i] = 0.5

# Distance from 0.5
domain_shift = np.abs(domain_auc - 0.5)

# ==================================================
# 3. COMBINED RANKING
# ==================================================

# Rank target signal:
# rank 1 = strongest target correlation
target_rank = pd.Series(
    -target_strength
).rank(method="average").values

# Rank domain stability:
# rank 1 = most stable
domain_rank = pd.Series(
    domain_shift
).rank(method="average").values

# Normalize ranks
target_rank_norm = target_rank / len(target_rank)
domain_rank_norm = domain_rank / len(domain_rank)

# Equal-weight combination
combined_score = (
    0.5 * target_rank_norm +
    0.5 * domain_rank_norm
)

ranking = np.argsort(combined_score)

print("\nTop 20 selected dimensions")
print("==========================")

for rank, idx in enumerate(ranking[:20], 1):
    print(
        f"{rank:2d}. wavlm_{idx:03d} | "
        f"corr={target_corr[idx]:+.4f} | "
        f"domain_auc={domain_auc[idx]:.4f}"
    )

# ==================================================
# 4. TEST FEATURE COUNTS
# ==================================================

train_idx, val_idx = train_test_split(
    np.arange(len(y)),
    test_size=0.20,
    random_state=42
)

y_tr = y[train_idx]
y_val = y[val_idx]

results = []

for k in [50, 100, 150, 200, 300, 400, 500, 600, 700]:

    selected = ranking[:k]

    # WavLM selected dimensions + all acoustic features
    X_selected = np.hstack([
        train_wavlm[:, selected],
        train_hand_X
    ])

    X_tr = X_selected[train_idx]
    X_val = X_selected[val_idx]

    scaler = StandardScaler()

    X_tr_scaled = scaler.fit_transform(X_tr)
    X_val_scaled = scaler.transform(X_val)

    model = Ridge(alpha=300)

    model.fit(X_tr_scaled, y_tr)

    pred = model.predict(X_val_scaled)

    rmse = np.sqrt(
        mean_squared_error(y_val, pred)
    )

    results.append((k, rmse))

    print(
        f"{k:3d} WavLM features + acoustic | "
        f"RMSE={rmse:.4f}"
    )

# ==================================================
# BEST
# ==================================================

best = min(results, key=lambda x: x[1])

print("\nBest:")
print("WavLM features:", best[0])
print("RMSE:", round(best[1], 4))
```

```python
import torch
import torchaudio
from torch.utils.data import Dataset, DataLoader

# --------------------------------------------------
# Fixed train/validation split
# --------------------------------------------------

from sklearn.model_selection import train_test_split

indices = np.arange(len(train))

train_idx, val_idx = train_test_split(
    indices,
    test_size=0.20,
    random_state=42
)

print("Training samples  :", len(train_idx))
print("Validation samples:", len(val_idx))


# --------------------------------------------------
# Audio Dataset
# --------------------------------------------------

class GrammarAudioDataset(Dataset):

    def __init__(self, dataframe, indices):
        self.dataframe = dataframe.iloc[indices].reset_index(drop=True)
        self.audio_dir = "/content/audio_train"

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, idx):

        row = self.dataframe.iloc[idx]

        filename = row["filename"]
        label = float(row["label"])

        path = os.path.join(
            self.audio_dir,
            filename
        )

        # Load audio
        waveform, sample_rate = torchaudio.load(path)

        # Convert stereo → mono if necessary
        if waveform.shape[0] > 1:
            waveform = waveform.mean(dim=0, keepdim=True)

        # Resample to 16 kHz
        if sample_rate != 16000:
            resampler = torchaudio.transforms.Resample(
                orig_freq=sample_rate,
                new_freq=16000
            )
            waveform = resampler(waveform)

        # Remove channel dimension
        waveform = waveform.squeeze(0)

        return waveform, torch.tensor(label, dtype=torch.float32)


# --------------------------------------------------
# Create datasets
# --------------------------------------------------

train_dataset = GrammarAudioDataset(
    train,
    train_idx
)

val_dataset = GrammarAudioDataset(
    train,
    val_idx
)

print("\nDatasets created successfully")

# Test one sample
audio, label = train_dataset[0]

print("\nFirst sample:")
print("Audio shape:", audio.shape)
print("Duration:", round(len(audio) / 16000, 2), "seconds")
print("Label:", label.item())
```

```python
import random

CROP_SECONDS = 20
CROP_SAMPLES = 16000 * CROP_SECONDS


def crop_audio(waveform, training=True):
    """
    Crop audio to a fixed 20-second segment.

    Training:
        Random crop for augmentation.

    Validation:
        Center crop for reproducibility.
    """

    length = waveform.shape[0]

    # If audio is shorter than 20 seconds,
    # pad with zeros.
    if length < CROP_SAMPLES:

        padding = CROP_SAMPLES - length

        waveform = torch.nn.functional.pad(
            waveform,
            (0, padding)
        )

        return waveform

    # Choose crop location
    if training:
        start = random.randint(
            0,
            length - CROP_SAMPLES
        )
    else:
        start = (length - CROP_SAMPLES) // 2

    return waveform[
        start:start + CROP_SAMPLES
    ]


def collate_fn(batch, training=True):

    waveforms = []
    labels = []

    for waveform, label in batch:

        waveform = crop_audio(
            waveform,
            training=training
        )

        waveforms.append(waveform)
        labels.append(label)

    waveforms = torch.stack(waveforms)
    labels = torch.stack(labels)

    return waveforms, labels


# Test batching
sample_batch = [
    train_dataset[i]
    for i in range(2)
]

waveforms, labels = collate_fn(
    sample_batch,
    training=True
)

print("Batch shape :", waveforms.shape)
print("Labels shape:", labels.shape)
print("Duration    :", waveforms.shape[1] / 16000, "seconds")
print("Labels      :", labels.tolist())
```

```python
import torch
import torch.nn as nn

from transformers import AutoFeatureExtractor, AutoModel

# --------------------------------------------------
# Device
# --------------------------------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

# --------------------------------------------------
# Load WavLM
# --------------------------------------------------

MODEL_NAME = "microsoft/wavlm-base-plus"

feature_extractor = AutoFeatureExtractor.from_pretrained(
    MODEL_NAME
)

wavlm = AutoModel.from_pretrained(
    MODEL_NAME
)

# --------------------------------------------------
# Regression model
# --------------------------------------------------

class WavLMGrammarModel(nn.Module):

    def __init__(self, wavlm_model):
        super().__init__()

        self.wavlm = wavlm_model

        self.regressor = nn.Sequential(
            nn.Linear(768, 256),
            nn.ReLU(),
            nn.Dropout(0.2),

            nn.Linear(256, 64),
            nn.ReLU(),
            nn.Dropout(0.1),

            nn.Linear(64, 1)
        )

    def forward(self, input_values):

        outputs = self.wavlm(
            input_values=input_values
        )

        # Mean pooling across time
        hidden = outputs.last_hidden_state

        pooled = hidden.mean(dim=1)

        prediction = self.regressor(
            pooled
        )

        return prediction.squeeze(-1)


model = WavLMGrammarModel(wavlm)

model = model.to(device)

# --------------------------------------------------
# Parameter information
# --------------------------------------------------

total_params = sum(
    p.numel()
    for p in model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print("\nModel created")
print("Total parameters    :", f"{total_params:,}")
print("Trainable parameters:", f"{trainable_params:,}")

# --------------------------------------------------
# GPU information
# --------------------------------------------------

if torch.cuda.is_available():

    print("\nGPU:", torch.cuda.get_device_name(0))

    print(
        "GPU memory:",
        round(
            torch.cuda.get_device_properties(0).total_memory
            / 1024**3,
            2
        ),
        "GB"
    )
```

```python
import time
import numpy as np
import torch

def evaluate_model(model, loader):
    model.eval()
    predictions = []
    targets = []

    with torch.no_grad():
        for audio, labels in loader:
            audio = audio.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            preds = model(audio)

            predictions.extend(preds.cpu().numpy())
            targets.extend(labels.cpu().numpy())

    predictions = np.array(predictions)
    targets = np.array(targets)

    rmse = np.sqrt(np.mean((predictions - targets) ** 2))
    return rmse, predictions, targets


# Train for one epoch
model.train()
model.wavlm.eval()

start_time = time.time()
train_losses = []

for batch_idx, (audio, labels) in enumerate(train_loader):

    audio = audio.to(device, non_blocking=True)
    labels = labels.to(device, non_blocking=True)

    optimizer.zero_grad(set_to_none=True)

    predictions = model(audio)

    loss = criterion(predictions, labels)

    loss.backward()
    optimizer.step()

    train_losses.append(loss.item())

    if (batch_idx + 1) % 50 == 0:
        print(
            f"Batch {batch_idx + 1}/{len(train_loader)} | "
            f"Loss: {loss.item():.4f}"
        )

elapsed = time.time() - start_time

# Validation
val_rmse, val_predictions, val_targets = evaluate_model(model, val_loader)

print("\n===== 1 EPOCH RESULT =====")
print("Average train loss:", round(np.mean(train_losses), 4))
print("Validation RMSE:", round(val_rmse, 4))
print("Time:", round(elapsed / 60, 2), "minutes")
```

```python
import os
import numpy as np
import torch
import torchaudio

# Layers we want to compare
SELECTED_LAYERS = [4, 8, 12]

layer_embeddings = {
    layer: np.zeros(
        (len(train), 768),
        dtype=np.float32
    )
    for layer in SELECTED_LAYERS
}

wavlm.eval()

print("Extracting layers:", SELECTED_LAYERS)
print("Training files:", len(train))
print()

for i, filename in enumerate(train["filename"]):

    path = os.path.join(
        "/content/audio_train",
        filename
    )

    waveform, sample_rate = torchaudio.load(path)

    # Mono
    if waveform.shape[0] > 1:
        waveform = waveform.mean(
            dim=0,
            keepdim=True
        )

    # 16 kHz
    if sample_rate != 16000:
        resampler = torchaudio.transforms.Resample(
            orig_freq=sample_rate,
            new_freq=16000
        )
        waveform = resampler(waveform)

    waveform = waveform.squeeze(0).unsqueeze(0).to(device)

    with torch.no_grad():
        outputs = wavlm(
            input_values=waveform,
            output_hidden_states=True
        )

    for layer in SELECTED_LAYERS:
        hidden = outputs.hidden_states[layer]

        # Mean pool over time
        embedding = hidden.mean(dim=1)

        layer_embeddings[layer][i] = (
            embedding.squeeze(0)
            .cpu()
            .numpy()
        )

    del waveform, outputs

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    if (i + 1) % 50 == 0:
        print(
            f"Processed {i + 1}/{len(train)}"
        )

print("\nExtraction complete")

for layer in SELECTED_LAYERS:
    print(
        f"Layer {layer}:",
        layer_embeddings[layer].shape
    )
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error

# ------------------------------------------------------------
# CORAL function
# Align training feature covariance to target/test covariance
# ------------------------------------------------------------

def coral_transform(X_source, X_target):
    eps = 1e-5

    # Center
    source_mean = X_source.mean(axis=0)
    target_mean = X_target.mean(axis=0)

    Xs = X_source - source_mean
    Xt = X_target - target_mean

    # Covariance matrices
    Cs = np.cov(Xs, rowvar=False) + eps * np.eye(Xs.shape[1])
    Ct = np.cov(Xt, rowvar=False) + eps * np.eye(Xt.shape[1])

    # Eigen decomposition
    eig_s, vec_s = np.linalg.eigh(Cs)
    eig_t, vec_t = np.linalg.eigh(Ct)

    eig_s = np.maximum(eig_s, eps)
    eig_t = np.maximum(eig_t, eps)

    Cs_inv_sqrt = (
        vec_s
        @ np.diag(1.0 / np.sqrt(eig_s))
        @ vec_s.T
    )

    Ct_sqrt = (
        vec_t
        @ np.diag(np.sqrt(eig_t))
        @ vec_t.T
    )

    # Source -> target covariance space
    transform = Cs_inv_sqrt @ Ct_sqrt

    X_source_aligned = Xs @ transform

    return X_source_aligned


# ------------------------------------------------------------
# Use the original random validation split
# ------------------------------------------------------------

X_source_train = train_wavlm_aligned[train_idx]
X_source_val = train_wavlm_aligned[val_idx]

# Test set is the target domain
X_target = test_wavlm_aligned

print("Source train:", X_source_train.shape)
print("Target test :", X_target.shape)

# CORAL
X_source_aligned = coral_transform(
    X_source_train,
    X_target
)

# Apply same transformation to validation data
source_mean = X_source_train.mean(axis=0)

X_val_centered = X_source_val - source_mean

# Reconstruct transformation
def coral_matrix(X_source, X_target):
    eps = 1e-5

    Cs = np.cov(
        X_source - X_source.mean(axis=0),
        rowvar=False
    ) + eps * np.eye(X_source.shape[1])

    Ct = np.cov(
        X_target - X_target.mean(axis=0),
        rowvar=False
    ) + eps * np.eye(X_target.shape[1])

    eig_s, vec_s = np.linalg.eigh(Cs)
    eig_t, vec_t = np.linalg.eigh(Ct)

    eig_s = np.maximum(eig_s, eps)
    eig_t = np.maximum(eig_t, eps)

    Cs_inv_sqrt = (
        vec_s
        @ np.diag(1 / np.sqrt(eig_s))
        @ vec_s.T
    )

    Ct_sqrt = (
        vec_t
        @ np.diag(np.sqrt(eig_t))
        @ vec_t.T
    )

    return Cs_inv_sqrt @ Ct_sqrt


transform = coral_matrix(
    X_source_train,
    X_target
)

X_val_aligned = X_val_centered @ transform

# ------------------------------------------------------------
# Standardize after alignment
# ------------------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_source_aligned
)

X_val_scaled = scaler.transform(
    X_val_aligned
)

# ------------------------------------------------------------
# Ridge
# ------------------------------------------------------------

for alpha in [100, 200, 300, 500]:

    model = Ridge(alpha=alpha)

    model.fit(
        X_train_scaled,
        y_tr
    )

    pred = model.predict(X_val_scaled)

    rmse = np.sqrt(
        mean_squared_error(y_val, pred)
    )

    print(
        f"alpha={alpha:3d} | RMSE={rmse:.4f}"
    )
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error
from sklearn.metrics.pairwise import cosine_similarity

# ------------------------------------------------------------
# 1. Remove the special zero-score population
# ------------------------------------------------------------

nonzero_mask = y > 0

X_train_base = train_wavlm_aligned[nonzero_mask]
y_train_base = y[nonzero_mask]

# ------------------------------------------------------------
# 2. Calculate how similar each training sample is to TEST
# ------------------------------------------------------------

# Normalize WavLM embeddings
train_norm = X_train_base / (
    np.linalg.norm(X_train_base, axis=1, keepdims=True) + 1e-12
)

test_norm = test_wavlm_aligned / (
    np.linalg.norm(test_wavlm_aligned, axis=1, keepdims=True) + 1e-12
)

# Similarity to every test recording
similarity_matrix = train_norm @ test_norm.T

# Average similarity to test population
test_similarity = similarity_matrix.mean(axis=1)

print("Similarity statistics:")
print("Min :", round(test_similarity.min(), 4))
print("Max :", round(test_similarity.max(), 4))
print("Mean:", round(test_similarity.mean(), 4))
print("Std :", round(test_similarity.std(), 4))

# ------------------------------------------------------------
# 3. Convert similarity into positive sample weights
# ------------------------------------------------------------

# Standardize similarity
z = (
    test_similarity - test_similarity.mean()
) / (test_similarity.std() + 1e-8)

# Try several weighting strengths
for strength in [0.0, 0.25, 0.5, 1.0, 2.0]:

    weights = np.exp(
        strength * z
    )

    # Normalize average weight to 1
    weights = weights / weights.mean()

    # --------------------------------------------------------
    # IMPORTANT:
    # Use the ORIGINAL validation split for fair comparison.
    # Only non-zero validation samples are evaluated.
    # --------------------------------------------------------

    train_nonzero_indices = np.where(
        nonzero_mask
    )[0]

    val_mask = np.isin(
        train_nonzero_indices,
        val_idx
    )

    # If validation samples are non-zero
    tr_indices = np.where(~val_mask)[0]
    va_indices = np.where(val_mask)[0]

    X_tr = X_train_base[tr_indices]
    X_val = X_train_base[va_indices]

    y_tr_weight = y_train_base[tr_indices]
    y_val_weight = y_train_base[va_indices]

    w_tr = weights[tr_indices]

    scaler = StandardScaler()

    X_tr_scaled = scaler.fit_transform(X_tr)
    X_val_scaled = scaler.transform(X_val)

    model = Ridge(alpha=300)

    model.fit(
        X_tr_scaled,
        y_tr_weight,
        sample_weight=w_tr
    )

    pred = model.predict(X_val_scaled)

    rmse = np.sqrt(
        mean_squared_error(
            y_val_weight,
            pred
        )
    )

    print(
        f"strength={strength:<4} | "
        f"nonzero validation RMSE={rmse:.4f}"
    )
```

```python
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error
import numpy as np

# ------------------------------------------------------------
# Features
# ------------------------------------------------------------

X_wavlm = np.hstack([
    train_wavlm_aligned,
    X_hand
])

X_hubert = np.hstack([
    train_hubert,
    X_hand
])

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

oof_wavlm = np.zeros(len(y))
oof_hubert = np.zeros(len(y))

# ------------------------------------------------------------
# 5-fold OOF predictions
# ------------------------------------------------------------

for fold, (tr_idx, va_idx) in enumerate(kf.split(X_wavlm), 1):

    # WavLM
    scaler_w = StandardScaler()

    Xtr_w = scaler_w.fit_transform(
        X_wavlm[tr_idx]
    )

    Xva_w = scaler_w.transform(
        X_wavlm[va_idx]
    )

    model_w = Ridge(alpha=300)

    model_w.fit(
        Xtr_w,
        y[tr_idx]
    )

    oof_wavlm[va_idx] = model_w.predict(
        Xva_w
    )

    # HuBERT
    scaler_h = StandardScaler()

    Xtr_h = scaler_h.fit_transform(
        X_hubert[tr_idx]
    )

    Xva_h = scaler_h.transform(
        X_hubert[va_idx]
    )

    model_h = Ridge(alpha=200)

    model_h.fit(
        Xtr_h,
        y[tr_idx]
    )

    oof_hubert[va_idx] = model_h.predict(
        Xva_h
    )

    print(f"Completed fold {fold}/5")


# ------------------------------------------------------------
# Individual OOF scores
# ------------------------------------------------------------

rmse_w = np.sqrt(
    mean_squared_error(y, oof_wavlm)
)

rmse_h = np.sqrt(
    mean_squared_error(y, oof_hubert)
)

print("\nOOF WavLM RMSE :", round(rmse_w, 4))
print("OOF HuBERT RMSE:", round(rmse_h, 4))


# ------------------------------------------------------------
# Search blend weights
# ------------------------------------------------------------

print("\nBlend results:")

best_rmse = float("inf")
best_weight = None

for w in np.arange(0.0, 1.01, 0.02):

    pred = (
        w * oof_wavlm
        + (1 - w) * oof_hubert
    )

    rmse = np.sqrt(
        mean_squared_error(y, pred)
    )

    if rmse < best_rmse:
        best_rmse = rmse
        best_weight = w

    if w in [0.0, 0.2, 0.4, 0.5, 0.6, 0.7, 0.8, 1.0]:
        print(
            f"WavLM weight={w:.2f} | "
            f"RMSE={rmse:.4f}"
        )

print("\nBEST:")
print("WavLM weight:", round(best_weight, 2))
print("HuBERT weight:", round(1 - best_weight, 2))
print("OOF RMSE:", round(best_rmse, 4))
```

```python
from sklearn.model_selection import KFold
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

raw_predictions = np.zeros(len(y))
calibrated_predictions = np.zeros(len(y))

for fold, (tr_idx, va_idx) in enumerate(kf.split(X_wavlm), 1):

    # --------------------------------------------------------
    # Train WavLM
    # --------------------------------------------------------

    scaler_w = StandardScaler()

    Xtr_w = scaler_w.fit_transform(
        X_wavlm[tr_idx]
    )

    Xva_w = scaler_w.transform(
        X_wavlm[va_idx]
    )

    model_w = Ridge(alpha=300)

    model_w.fit(
        Xtr_w,
        y[tr_idx]
    )

    pred_w_tr = model_w.predict(Xtr_w)
    pred_w_va = model_w.predict(Xva_w)

    # --------------------------------------------------------
    # Train HuBERT
    # --------------------------------------------------------

    scaler_h = StandardScaler()

    Xtr_h = scaler_h.fit_transform(
        X_hubert[tr_idx]
    )

    Xva_h = scaler_h.transform(
        X_hubert[va_idx]
    )

    model_h = Ridge(alpha=200)

    model_h.fit(
        Xtr_h,
        y[tr_idx]
    )

    pred_h_tr = model_h.predict(Xtr_h)
    pred_h_va = model_h.predict(Xva_h)

    # --------------------------------------------------------
    # 66/34 blend
    # --------------------------------------------------------

    blend_tr = (
        0.66 * pred_w_tr
        + 0.34 * pred_h_tr
    )

    blend_va = (
        0.66 * pred_w_va
        + 0.34 * pred_h_va
    )

    raw_predictions[va_idx] = blend_va

    # --------------------------------------------------------
    # Learn calibration ONLY from training fold
    # --------------------------------------------------------

    calibrator = LinearRegression()

    calibrator.fit(
        blend_tr.reshape(-1, 1),
        y[tr_idx]
    )

    calibrated_predictions[va_idx] = calibrator.predict(
        blend_va.reshape(-1, 1)
    )

    print(
        f"Fold {fold}: "
        f"slope={calibrator.coef_[0]:.4f}, "
        f"intercept={calibrator.intercept_:.4f}"
    )


# ------------------------------------------------------------
# Final leakage-safe scores
# ------------------------------------------------------------

raw_predictions = np.clip(
    raw_predictions,
    0,
    5
)

calibrated_predictions = np.clip(
    calibrated_predictions,
    0,
    5
)

raw_rmse = np.sqrt(
    mean_squared_error(
        y,
        raw_predictions
    )
)

calibrated_rmse = np.sqrt(
    mean_squared_error(
        y,
        calibrated_predictions
    )
)

print("\n==============================")
print("Leakage-safe OOF results")
print("==============================")

print(
    "Raw blend RMSE       :",
    round(raw_rmse, 4)
)

print(
    "Calibrated blend RMSE:",
    round(calibrated_rmse, 4)
)

print(
    "\nAverage calibration slope:",
    round(
        np.mean([
            1.0  # placeholder; fold slopes printed above
        ]),
        4
    )
)
```

```python
import os
import time
import numpy as np
import pandas as pd
import torch
import librosa

# Paths
train_audio_dir = "/content/audio_train"
checkpoint_dir = "/content/drive/MyDrive/SHL_Grammar/checkpoints"
os.makedirs(checkpoint_dir, exist_ok=True)

train_csv = "/content/drive/MyDrive/SHL_Grammar/train.csv"

train_df = pd.read_csv(train_csv)

# IMPORTANT: follow CSV order
train_files = train_df["filename"].tolist()

train_embeddings = []

start_time = time.time()

for i, filename in enumerate(train_files):

    audio_path = os.path.join(train_audio_dir, filename)

    waveform, sr = librosa.load(
        audio_path,
        sr=16000,
        mono=True
    )

    inputs = wav2vec_extractor(
        waveform,
        sampling_rate=16000,
        return_tensors="pt"
    )

    inputs = {
        k: v.to(device)
        for k, v in inputs.items()
    }

    with torch.no_grad():
        outputs = wav2vec_model(**inputs)

    # Mean pooling over time
    embedding = outputs.last_hidden_state.mean(dim=1)

    train_embeddings.append(
        embedding.squeeze(0).cpu().numpy()
    )

    if (i + 1) % 50 == 0:
        elapsed = time.time() - start_time
        print(
            f"{i+1}/{len(train_files)} "
            f"| elapsed: {elapsed/60:.1f} min"
        )

train_w2v = np.stack(train_embeddings)

save_path = os.path.join(
    checkpoint_dir,
    "train_wav2vec2_aligned.npy"
)

np.save(save_path, train_w2v)

print("\nDONE")
print("Shape:", train_w2v.shape)
print("NaNs:", np.isnan(train_w2v).sum())
print("Saved:", save_path)
print("Total time:", round((time.time()-start_time)/60, 2), "minutes")
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error

# ------------------------------------------------------------
# Load labels and embeddings
# ------------------------------------------------------------

train_df = pd.read_csv(
    "/content/drive/MyDrive/SHL_Grammar/train.csv"
)

y = train_df["label"].values

X_wavlm = np.load(
    "/content/drive/MyDrive/SHL_Grammar/checkpoints/train_wavlm_aligned.npy"
)

X_hubert = np.load(
    "/content/drive/MyDrive/SHL_Grammar/checkpoints/train_hubert_aligned.npy"
)

X_w2v = np.load(
    "/content/drive/MyDrive/SHL_Grammar/checkpoints/train_wav2vec2_aligned.npy"
)

# ------------------------------------------------------------
# Same fixed validation split
# ------------------------------------------------------------

train_idx, val_idx = train_test_split(
    np.arange(len(y)),
    test_size=0.20,
    random_state=42
)

def ridge_predict(X, alpha):
    scaler = StandardScaler()

    X_train = scaler.fit_transform(X[train_idx])
    X_val = scaler.transform(X[val_idx])

    model = Ridge(alpha=alpha)

    model.fit(
        X_train,
        y[train_idx]
    )

    return model.predict(X_val)

# WavLM
pred_wavlm = ridge_predict(
    X_wavlm,
    300
)

# HuBERT
pred_hubert = ridge_predict(
    X_hubert,
    200
)

# Wav2Vec2
pred_w2v = ridge_predict(
    X_w2v,
    500
)

y_val = y[val_idx]

# ------------------------------------------------------------
# Blend WavLM + HuBERT first
# ------------------------------------------------------------

base_pred = (
    0.66 * pred_wavlm
    + 0.34 * pred_hubert
)

base_rmse = np.sqrt(
    mean_squared_error(y_val, base_pred)
)

print("WavLM + HuBERT RMSE:", round(base_rmse, 4))

# ------------------------------------------------------------
# Add small amounts of Wav2Vec2
# ------------------------------------------------------------

for w in [0.02, 0.05, 0.10, 0.15, 0.20, 0.25]:

    blended = (
        (1 - w) * base_pred
        + w * pred_w2v
    )

    rmse = np.sqrt(
        mean_squared_error(y_val, blended)
    )

    print(
        f"Wav2Vec2 weight={w:.2f} | "
        f"RMSE={rmse:.4f}"
    )
```
